Bài 1: NHập vào một số nguyên dương có 3 chữ số . Tìm và in ra màn hình "SỐ NGUYÊN " đảo ngược của n . KHông dunngf kiểu chuỗi . chú ý :Kiểm tra chỉ xử lý các số nguyên dương có 3 chữ số . Các số khác không xử lý

```python
def kiem_tra_ba_so(n):
    cnt = 0
    if n <= 0:
        return False
    else:
        while n > 0:
           
           if n % 10 > 0:
            cnt += 1
            n = n // 10
        if cnt == 3:
            return True
        else:
            return False
def dao_nguoc(a):
    so_dao = 0
    while a > 0:
        so_dao = so_dao * 10 + a % 10
        a = a // 10
    print(so_dao)
b = int(input())
if kiem_tra_ba_so(b):
    dao_nguoc(b)
```

Bài 4 : Viết chương trình in ra hình tam giác cân có đọ cao h đc nhập từ bàn phím

```python
n = int(input())
for i in range(1 , n + 1):
    if i == 1:
        print(" " * (n - 1) + "*")
    elif i == n:
        print("*" * (2 * n - 1))
    else:
        khoang_trang_truoc = (n - i) * " "
        khoang_trang_giua = (2 * i - 3) * " "
        print(khoang_trang_truoc + "*" + khoang_trang_giua + "*")
```
