Bài 3: Viết chương trình nhập vào 1 kí tự c, nếu nó là chữ hoa thì in ra chữ thường tương ứng ngược lại nếu nó là chững thường thì in ra chữ hoa tương ứng của nó

```python
def kiem_tra(kitu):
    if kitu.isupper():
        print(kitu.lower())
    else:
        print(kitu.upper())
bien = input("Nhập vào một kí tự: ")
kiem_tra(bien)
```

Bài 4 : Viết chương trình nhập vào một kí tự , kiểm tra xem nó có phải kí tự chữ cái không (hoa or thường). Nếu đúng in ra 'c' là kí tự alphabet ,nếu sai in ra 'c' không là kí tự alphabet

```python
ki_tu = input("Nhập vào kí tự: ")
if ki_tu.isalpha():
    print(f"{ki_tu} là kí tự alphabet")
else:
    print(f"{ki_tu} không phải là ki tự alphabet")
```

Bài 5: Nhập vào chữ cái in hoa A , in ra chữ cái thường liền trước chữ cái a thường tương ứng của A . Xét trường hợp khi c = 'A' là đặc biệt

```python
c = input("Nhập vào chữ cái in hoa: ")
if c == 'A' : ## gắn giá trị kèm theo kí tự ''
    print('z')
else:
    print(chr(ord(c.lower())-1))
```

```python
ten_chu_ho = input("Ten chu ho: ")
chi_so_truoc = int(input("Chi so thang truoc: "))
chi_so_nay = int(input("Chi so thang nay: "))

so_dien = chi_so_nay - chi_so_truoc
tien_dien = 0

if so_dien <= 50:
    tien_dien = so_dien * 1984
elif so_dien <= 100:
    tien_dien = (50 * 1984) + ((so_dien - 50) * 2050)
elif so_dien <= 200:
    tien_dien = (50 * 1984) + (50 * 2050) + ((so_dien - 100) * 2380)
elif so_dien <= 300:
    tien_dien = (50 * 1984) + (50 * 2050) + (100 * 2380) + ((so_dien - 200) * 2998)
elif so_dien <= 400:
    tien_dien = (50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + ((so_dien - 300) * 3350)
else:
    tien_dien = (50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + (100 * 3350) + ((so_dien - 400) * 3460)

tong_tien_vat = round(tien_dien * 1.08)

print(f"Ho va ten: {ten_chu_ho}")
print(f"Tien phai tra la: {tong_tien_vat}")
```

Bài 6: Nhập vào 3 số a, b, c. kiểm tra xem đây có phải là 1 cạnh của tam giác không nếu đúng thì in ra diện tích của tam giác đó , in ra với 1 chữ số phần thập phân , nếu nó không phải 3 cạnh của tam giác thì in ra "Khong phai 3 canh tam giac"

```python
import math
def kiem_tra_tam_giac( a , b , c):
    lst = [ a , b , c]
    lst.sort()
    if lst[2] < lst[1] + lst[0]:
        return True

x , y , z = map(float , input().split())
p = (x + y + z)/2
s = math.sqrt(p * (p - x) * (p - y) * (p - z))
if kiem_tra_tam_giac(x , y , z):
    print(s)
else:
    print("Khong la 3 canh tam giac")
```
