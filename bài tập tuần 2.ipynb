Bài 3:Tính tam giác .Nhập vào ba số a,b,c bất kỳ. Kiểm tra xem nó có thể là độ dài ba cạnh hay không ,nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'.NGược lại ,kiem tra xem no la tam giác j : Đều , Cân , Vuông hay tam giác thường . tính diện tích , chu vi của tam giác và in ra màn hình

```python
import math
def kiem_tra( a , b , c):
    lst = [ a , b , c]
    lst.sort()
    if lst[0] + lst[1] > lst[2]:
        return True
x , y , z = map(float , input().split())
c = [x , y , z ]
c.sort()
if kiem_tra(x , y , z) == False:
    print("Khong la ba canh cua mot tam giac")
else:
    p = (x + y + z) / 2
    s = math.sqrt(p * (p - x) * (p - y) * (p - z))
    if x == y == z:
        print(f"Tam giac Đeu {s:.2f} , {p}")
    elif c[0] == c[1] or c[1] == c[2]: 
        print(f"Tam giac Can {s:.2f},{p}")
    elif c[0]**2 + c[1]**2 ==c[2]**2:
        print(f"Tam giac Vuong {s:.2f} , {p}")
    else:
        print(f"Tam giac thuong {s:.2f},{p}")
```

Bài 5: Viết chương trình giải phương trình bậc 2

```python
import math
def tinh_toan( a , b ,c ):
    delta = b**2 - 4 * a * c

    if delta == 0:
        print(f"Phuong trình =có ngiệm kép là { - b / 2 * a :.1f}")
    elif delta < 0:
        print("Phương trình vô nghiệm")
    else:
        x1 = (- b + math.sqrt(delta)) / 2
        x2 = (- b - math.sqrt(delta)) / 2
        print(f"{x1:.1f} , {x2:.1f}")
x , y , z = map(float , input().split())
tinh_toan(x , y ,z)
```

Bài 6 : TÌm giá trị lớn nhất trong bốn số a , b, c ,d được nhập từ bàn phím

```python
a , b , c , d = map(int , input().split())
max = a
if max < b:
    max = b
if max < c:
    max = c
if max < d:
    max = d
print(max)
```

Bài 10: Nhập vào tâm và bán kính của 1 đường tròn . Sau đó nhập vào 1 điểm A(x , y) bất kỳ xem nó có thuộc đường tròn hay không

```python
import math
a , b , R = map(float , input().split())
x , y = map(float , input().split())
len = math.sqrt((x - a)**2 + (y - b)**2)
if len == R:
    print("True")
else:
    print("No")
```

```python
# cách 1:
x , y = map(float , input().split())
cnt = 1
tich = 1
while cnt <= y:
    tich = tich  * x
    cnt += 1
print(tich)
```
